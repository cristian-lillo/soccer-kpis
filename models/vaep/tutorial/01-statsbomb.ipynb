## Load Data

```python
# Initialize StatsBomb data loader
SBL = StatsBombLoader(
    getter="local",
    root=str(paths.STATSBOMB_DIR),
)
```

```python
# Load the list of competitions
competitions_df = SBL.competitions()
competitions_df
```

```python
# Define a specific competition and season to work with (e.g., UEFA Euro 2024)
tournament = tournaments.EURO_2024
competition_id = tournament["competition_id"]
season_id = tournament["season_id"]
```

```python
# Load the games data for the specified competition and season
games_df = SBL.games(
    competition_id=competition_id,
    season_id=season_id,
)

games_df.head()
```

```python
# Load the teams data for the specific match
teams_df = SBL.teams(game_id=tournaments.EURO_2024_FINAL_MATCH_ID)
teams_df
```

```python
# Load the players data for the specific match
players_df = SBL.players(game_id=tournaments.EURO_2024_FINAL_MATCH_ID)
players_df
```

```python
# Load the events data for the specific match
events_df = SBL.events(game_id=tournaments.EURO_2024_FINAL_MATCH_ID)
events_df.head()
```

```python
# Load the 360 data for the specific match
events_360_df = SBL.events(game_id=tournaments.EURO_2024_FINAL_MATCH_ID, load_360=True)
events_360_df.head()
```

## SPADL

```python
# Get the home team ID for the specific match
home_team_id = games_df[games_df["game_id"] == tournaments.EURO_2024_FINAL_MATCH_ID]["home_team_id"].iloc[0]
```

```python
# Convert the events data of the specific match to SPADL actions
actions_df = spadl.statsbomb.convert_to_actions(events_df, home_team_id=home_team_id)
actions_df
```

```python
# Replace IDs with names in the actions DataFrame
actions_df = (
    spadl.add_names(actions_df)
    .merge(SBL.teams(game_id=tournaments.EURO_2024_FINAL_MATCH_ID))
    .merge(SBL.players(game_id=tournaments.EURO_2024_FINAL_MATCH_ID))
)
actions_df
```

## Number of matches

```python
n_matches_list = []

for _, row in competitions_df.iterrows():
    season_id = row["season_id"]
    competition_id = row["competition_id"]

    # Load the matches for the given competition and season
    matches_df = SBL.games(competition_id=competition_id, season_id=season_id)
    n_matches = len(matches_df)
    n_matches_list.append(n_matches)

competitions_with_games_df = competitions_df.copy()
competitions_with_games_df.insert(4, "n_matches", n_matches_list)
```

```python
print(competitions_with_games_df.to_string())
```

```python
three_sixty_matches = [int(p.stem) for p in paths.STATSBOMB_THREE_SIXTY_DIR.glob("*.json")]
```

```python
n_matches_360_list = []

for _, row in competitions_with_games_df.iterrows():
    competition_id = row["competition_id"]
    season_id = row["season_id"]

    matches_df = SBL.games(competition_id=competition_id, season_id=season_id)
    match_ids = matches_df["game_id"].tolist()
    n_matches_360 = sum(1 for match_id in match_ids if match_id in three_sixty_matches)
    n_matches_360_list.append(n_matches_360)

competitions_with_360_games_df = competitions_with_games_df.copy()
competitions_with_360_games_df.insert(5, "n_matches_360", n_matches_360_list)
```

```python
print(competitions_with_360_games_df.to_string())
```
