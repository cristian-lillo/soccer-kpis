# Socceraction

## Libraries and Constants

```python
# Import libraries
import socceraction.spadl as spadl

from pathlib import Path
from socceraction.data.statsbomb import StatsBombLoader
```

```python
# Ignore warnings
import warnings

warnings.filterwarnings("ignore", category=FutureWarning)
warnings.filterwarnings("ignore", message="Inferred xy_fidelity_version=2.", category=UserWarning)
```

```python
# Define the data directory
DATA_DIR = Path("../data")
```

## StatsBomb API

```python
# Define the StatsBomb API as local
statsbomb_data_dir = str(DATA_DIR)
SBL = StatsBombLoader(getter="local", root=statsbomb_data_dir)
```

```python
# Load the competitions data
competitions_df = SBL.competitions()
competitions_df
```

```python
# Load the games data for a specific competition and season
# For example, the  UEFA Euro 2024
games_df = SBL.games(competition_id=55, season_id=282).set_index("game_id")
games_df.sort_values(by="game_day", ascending=False, inplace=True)
games_df.head(n=7)
```

```python
# Load the teams data for the final game of the UEFA Euro 2024
teams_df = SBL.teams(game_id=3943043)
teams_df
```

```python
# Load the players data for the final game of the UEFA Euro 2024
players_df = SBL.players(game_id=3943043)
players_df
```

```python
# Load the events data for the final game of the UEFA Euro 2024
events_df = SBL.events(game_id=3943043)
events_df.head()
```

```python
# Load the 360 data for the final game of the UEFA Euro 2024
events_360_df = SBL.events(game_id=3943043, load_360=True)
events_360_df.head()
```

## SPADL

```python
# Convert the events data of the final game of the UEFA Euro 2024 to SPADL actions
home_team_id = games_df.at[3943043, "home_team_id"]
actions_df = spadl.statsbomb.convert_to_actions(events_df, home_team_id=home_team_id)
actions_df
```

```python
# Replace IDs with names in the actions DataFrame
actions_df = spadl.add_names(actions_df).merge(SBL.teams(game_id=3943043)).merge(SBL.players(game_id=3943043))
actions_df
```
